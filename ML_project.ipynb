#Data Cleaning (consolidated, single clean block)

1. Skewness check (quantifies what the boxplots showed)

```python
num_cols = ["person_age","person_income","person_emp_exp","loan_amnt",
            "loan_int_rate","loan_percent_income","cb_person_cred_hist_length","credit_score"]

skew_vals = df[num_cols].skew().sort_values(ascending=False)
print(skew_vals)
# Rule of thumb: |skew| > 1 -> highly skewed, consider log-transform
```

2. Fix impossible/error values (hard filter, not statistical outliers)

```python
# Already have age/emp_exp filtering — keep it, just consolidating here
df = df[df['person_age'] <= 100]
df = df[df['person_emp_exp'] <= (df['person_age'] - 14)]
df = df.reset_index(drop=True)
```

3. Winsorize (cap) genuinely skewed-but-real columns — don't drop rows

```python
def winsorize(series, lower_pct=0.01, upper_pct=0.99):
    lower = series.quantile(lower_pct)
    upper = series.quantile(upper_pct)
    return series.clip(lower, upper)

cap_cols = ['person_income', 'loan_amnt', 'person_emp_exp', 'cb_person_cred_hist_length']

for col in cap_cols:
    df[col] = winsorize(df[col])

# quick before/after check
df[cap_cols].describe()
```

4. Log-transform highly skewed columns (alternative/complement to capping — creates new columns so you don't lose the original)

```python
log_cols = ['person_income', 'person_emp_exp']

for col in log_cols:
    df[f'{col}_log'] = np.log1p(df[col])

# compare skew before vs after
for col in log_cols:
    print(f"{col}: original skew={df[col].skew():.2f}, log skew={df[f'{col}_log'].skew():.2f}")
```

5. Visual before/after comparison (confirms the capping/transform worked)


```python
fig, axes = plt.subplots(2, 2, figsize=(12,8))

sns.boxplot(y=df['person_income'], ax=axes[0,0]).set_title("person_income (capped)")
sns.boxplot(y=df['person_income_log'], ax=axes[0,1]).set_title("person_income (log)")
sns.boxplot(y=df['loan_amnt'], ax=axes[1,0]).set_title("loan_amnt (capped)")
sns.histplot(df['person_income_log'], kde=True, ax=axes[1,1]).set_title("person_income_log distribution")

plt.tight_layout()
plt.show()
```

6. Check whether outliers correlate with the target before finalizing (important — avoid silently biasing your model)

```python
# were the removed/capped rows disproportionately one class?
outlier_mask = (df['person_income'] == df['person_income'].max()) | (df['loan_amnt'] == df['loan_amnt'].max())
print(df.loc[outlier_mask, 'loan_status'].value_counts(normalize=True))
print(df['loan_status'].value_counts(normalize=True))
# if these differ a lot, outliers carry signal — be cautious about aggressive removal
```

7. Multicollinearity check (VIF) — useful if you're heading toward logistic regression

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

X = df[num_cols].dropna()
vif_data = pd.DataFrame()
vif_data["feature"] = X.columns
vif_data["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
print(vif_data.sort_values("VIF", ascending=False))
# VIF > 5-10 suggests strong multicollinearity
```

1. Encode categorical variables
Your categorical columns: person_gender, person_education, person_home_ownership, loan_intent, previous_loan_defaults_on_file.

```python
# Label encode binary/ordinal ones
df['person_gender'] = df['person_gender'].map({'male': 0, 'female': 1})
df['previous_loan_defaults_on_file'] = df['previous_loan_defaults_on_file'].map({'No': 0, 'Yes': 1})

# Education is ordinal - map it in order rather than one-hot
edu_order = {'High School': 0, 'Associate': 1, 'Bachelor': 2, 'Master': 3, 'Doctorate': 4}
df['person_education'] = df['person_education'].map(edu_order)

# One-hot encode nominal (no inherent order) categories
df = pd.get_dummies(df, columns=['person_home_ownership', 'loan_intent'], drop_first=True)
```
